До цього місяця, ми мали продажі в середньому на чек $20.

Необхідно зʼясувати, чи вийшло у нас статистично значущо **підвищити** середній чек за липень 2025?

Аби дати відповідь - ми проведемо z-test на рівні значущості $\alpha=0.05$ двома способами. В цьому завданні ми будемо виконувати обчислення "вручну" з використанням бібліотек numpy та scipy.stats подібно до прикладу в лекції. Для цього виконайте наступні 7 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть параметри задачі у змінні Python:
    - `sample_mean` - $\bar{x}$, середнє значення в вибірці
    - `population_mean` - $\mu_0$, середнє значення в популяції (тобто те, з яким ми порівнюємо середнє вибіркове значення)
    - `population_std` - $\sigma$, вибіркове стандартне відхилення, яке ми вважаємо, що є рівним ст. відх. популяції, адже маємо достатньо велику вибірку.
    - `sample_size` - $n$, розмір вибірки
    - `alpha` - $\alpha$ рівень значущості
    
    **Увага!** Для обчислення стандартного відхилення маємо скористатись функцією `np.std(your_dataframe, ddof=1)`. Чому так - розбираємо в лекції про t-test.


2. Визначте для цієї задачі:
    - якою є гіпотеза $H_0$
    - якою є альтернативна гіпотеза $H_a$
    - з яким типом тесту ми маємо справу - лівосторонній, правосторонній чи двосторонній.

3. Обчисліть стандартну помилку SE.
4. Розрахуйте z-статистику.
5. Знайдіть критичне z-значення з допомогою бібліотеки stats.
6. Обчисліть p-value з допомогою бібліотеки stats.
7. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$. Для прийняття рішення зробіть і порівняння z-статистики з критичним значенням, і проаналізуйте p-value.

```python
# 1. Запис параметрів у змінні python  
sample_mean =  data.mean()              # x̄ - середнє значення в вибірці
population_mean = 20                    # μ₀ - середнє значення в популяції (тобто те, з яким ми порівнюємо середнє вибіркове значення)
population_std = np.std(data,ddof=1)    # σ - вибіркове стандартне відхилення, яке ми вважаємо, що є рівним ст. відх. популяції, адже маємо достатньо велику вибірку.
sample_size = len(data)                 # n - розмір вибірки
alpha = 0.05                            # α - рівень значущості
```

2. Формулювання гіпотез

Гіпотези

H₀: μ = 20 (середній чек в липні 2025 не зріс)
H₁: μ > 20 (середній чек в липні 2025 зріс)
Тип тесту: правосторонній

Альтернативна гіпотеза ">”, тому тест правосторонній. Ми перевіряємо чи чек саме збільшився, а не просто змінився

```python
# 3.Обчислення стандартної помилки SE
# Формула: SE = σ / √n

standard_error = population_std / np.sqrt(sample_size)

print(f"Стандартна помилка = {standard_error:.3f} грошових одиниць чеку")
print("Інтерпретація: В середньому, вибіркове середнє відхиляється")
print(f"від справжнього середнього на ±{standard_error:.3f} грошових одиниць чеку")
```

```python
# 4. Розрахунок z-статистики
# Формула: z = (x̄ - μ₀) / SE

z_statistic = (sample_mean - population_mean) / standard_error

print(f"z-статистика: z = {z_statistic:.3f}")
print(f"Інтерпретація: вибіркове середнє відхиляється на {z_statistic:.3f} стандартних помилок")
print("від гіпотетичного середнього популяції")

```

```python
# 5. Знаходимо z-критичне значення
z_critical = stats.norm.ppf(1 - alpha)

print(f"Для рівня значущості α = {alpha} (правосторонній тест)")
print(f'z-критичне = {z_critical:.3f}')
```

```python
# 6.Обчислюємо p-value
p_value = 1 - stats.norm.cdf(z_statistic)
print(f'p-value = {p_value:.3f}')
```

```python
# 7. Приймаємо рішення. Метод 1
print('Метод 1. Порівняння z-статистики з критичним значенням')
if z_statistic > z_critical:
    print(f"z-статистика ({z_statistic:.3f}) > z-критичне ({z_critical:.3f})\nВисновок: z-статистика перевищує критичне значення")
    decision1 = "Відхиляємо гіпотезу H₀"
else:
    print(f'z-статистика ({z_statistic:.3f}) ≤ z-критичне ({z_critical:.3f})\nВисновок: z-статистика не перевищує критичне значення')
    decision1 = "Не відхиляємо гіпотезу H₀"
print(f'Рішення: {decision1}')

```

```python
# 7. Приймаємо рішення. Метод 2
print('Метод 2. Порівняння p-value з рівнем значущості ')

print(f'Якщо H₀ правдива, то ймовірність отримати вибіркове середнє ≥ {sample_mean:.2f} становить {p_value:.3f}')
if p_value < alpha:
    print("Висновок: p-value МЕНШЕ за рівень значущості \nРішення: Відхиляємо гіпотезу H₀")
else:
    print("Висновок: p-value БІЛЬШЕ за рівень значущості \nРішення: Не відхиляємо гіпотезу H₀")
```

```python
# 7. Приймаємо остаточне рішення
print('Заключення:')

if decision1 == "Відхиляємо гіпотезу H₀":
    print("   Середній чек в липні 2025 року вдалося збільшити на рівні значущості 5%:")
    print(f"   середній чек у липні ({sample_mean:.2f}) статистично значуще")
    print(f"   перевищує попередній середній чек ({population_mean})")
else:
    print("   Немає достатніх статистичних доказів, що середній чек в липні 2025 року вдалося збільшити:")
    print(f"   середній чек у липні ({sample_mean:.2f}) не дає підстав стверджувати,")
    print(f"   що він вищий за попередній середній чек ({population_mean})")
    if sample_mean < population_mean:
        print("   Більше того, вибіркове середнє навіть нижче за 20, тож правосторонній тест")
        print("   не міг показати статистично значуще зростання")
```

Чи зійшлись значення z-статистику та р-значення в цьому завданні з попередніми обчисленнями?

```python
from statsmodels.stats import weightstats as stests

# Виконуємо правосторонній z-тест
z_stat_sm, p_val_sm = stests.ztest(data, value=population_mean, alternative='larger')   

print(f"Результати z-тесту:")
print(f"  Z-статистика: {z_stat_sm:.3f}")
print(f"  P-value: {p_val_sm:.3f}")

# Інтерпретація результатів
print('Порівняння P-value з рівнем значущості:')

print(f' Якщо H₀ правдива, то ймовірність отримати вибіркове середнє ≥ {sample_mean:.2f} становить {p_val_sm:.3f}')
if p_val_sm < alpha:
    print(f" Висновок: P-value МЕНШЕ за рівень значущості {alpha} \n Рішення: Відхиляємо гіпотезу H₀")
else:
    print(f" Висновок: P-value БІЛЬШЕ за рівень значущості {alpha*100}% \n Рішення: Не відхиляємо гіпотезу H₀")

# Порівняння з ручними розрахунками

print('\nПорівняння з ручними розрахунками:')
print(f"Z-статистика:")
print(f"  Ручний розрахунок: {z_statistic:.3f}")
print(f"  Statsmodels:       {z_stat_sm:.3f}")
print(f"  Різниця:           {abs(z_statistic - z_stat_sm):.3f}")

print(f"Р-значення:")
print(f"  Ручний розрахунок: {p_value:.3f}")
print(f"  Statsmodels:       {p_val_sm:.3f}")
print(f"  Різниця:           {abs(p_value - p_val_sm):.3f}")

print("Різниці в результатах немає, отже можна використовувати обидва методи.")

```
