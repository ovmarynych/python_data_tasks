#Питання:

1) які категорії товарів приносять найбільший прибуток, а які найменший?

2) які категорії товарів найбільше купують жінки, а які чоловіки?

```python
# 1 питання: які категорії товарів приносять найбільший прибуток, а які найменший?

income_by_category = df.groupby(['Product line'])['gross income'].sum()
income_by_category.round()
```

Найбільший прибуток приносять Food and beverages, майже такий самий прибуток має категорія Sports and travel.

Найменший прибуток приносять Health and beauty.

Така інформація може бути корисною для прийняття рішень щодо категорій товарів, які потрібно й надалі підтримувати і розвивати, а на які звернути більше уваги та розробити маркетингову стратегію їх просування, або підкоригувати асортимент, або можливо й зняти з продажу.

В нашому випадку, наприклад, Health and beauty є кандидатом на більш ґрунтовний аналіз, щоби знайти можливу причину непопулярності серед покупців і вирішити як діяти, щоби збільшити її популярність.

Зняття з продажу є крайньою мірою, бо часто для супермаркетів важлива різноманітність товарів. і якщо наприклад прийняти рішення зняти з продажу категорію Health and beauty, то людина ймовірніше вибере інший супермаркет, щоби зробити свої регулярні покупки, де є ця категорія товарів в продажу.

```python
# 2 питання: які категорії товарів найбільше купують жінки, а які чоловіки?
revenue_by_gender = pd.pivot_table(df,index="Product line", columns='Gender', values='Total', aggfunc='sum')
revenue_by_gender.loc['Grand Total'] = revenue_by_gender.sum(numeric_only=True).values
revenue_by_gender.round()
```

Жінки найбільше зробили покупок із категорії Food and beverages, а чоловіки - Health and beauty(що є неочікувано).

Наступні по популярності у жінок категорії: Fashion accessories та Home and lifestyle. Популярність Food and beverages, Fashion accessories та Home and lifestyle у жінок є цілком очікуваною.

У чоловіків також популярні категорії: Electronic accessories та Sports and travel.

Найменш популярні категорії у жінок - Health and beauty, а чоловіків - Food and beverages.

Головний інсайт - категорія Health and beauty, яка є найпопулярнішою у чоловіків і найменш популярною серед жінок.

Такв інформація може використовуватися для більш таргетованих пропозицій: різних для жінок та чоловіків, а також для аналізу та можливому перегляду асортименту товарів. Наприклад, щодо нашого інсайту треба зрозуміти чому чоловіки купують товари із Health and beauty більше за жінок. Тут може схований суттєвий додатковий ріст продажів.

```python
# 3 питання: які методи оплати використовують найбільше?
df.groupby(['Payment'])['Total'].agg(['count','sum','mean'])
```

Найчастіше покупці використовують метод оплати Ewallet (345 разів) та майже так само Cash (344 рази).
Але суми оплачених товарів готівкою вище, ніж Ewallet (112.2 тис иа 110 тис відповідно).
Найбільша середня оплата у метода Cash (326.18), а у метода Ewallet вона найменша (318.82)

Така інформація дає можливість бізнесу прийняти рішення щодо підтримання/ не підтримання методів оплат, а також щодо розвитку окремих методів оплат для збільшення виручки та прибутку.

В нашому випадку є доречним звернути увагу на метод Ewallet, який використовується так само часто, але має найменшу середню оплату. Але тут важливо обережно ставитись до висновків, щоби наприклад за додаткові бонуси за розрахунок Ewallet не сталося просте перетікання з одного методу оплати на інший, не створюючи додаткової вартості бізнесу.
