# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
import pandas as pd
import numpy as np

from scipy import stats
```

```python
# дані з умови
times = np.array([128, 135, 121, 142, 126, 151, 123])
mu0 = 123.7
alpha = 0.1
```

```python
 # 1. Формулювання гіпотез
print('Гіпотези:')
print(f'H_0: μ = {mu0} хвилин (середній час до покупки користувачів не зріс)')
print(f'H_1: μ > {mu0} хвилин (середній час до покупки користувачів зріс)')
print('Тип тесту: правосторонній')
print('Альтернативна гіпотеза ">”, тому тест правосторонній. Ми перевіряємо чи час саме збільшився, а не просто змінився')
```

```python
# 2. Обчислення вибіркової статистики

x_bar = np.mean(times)
s = np.std(times, ddof=1)
n= len(times)
print(f'Вибіркове середнє:  {x_bar:.3f}')
print(f'Вибіркове стандартне відхилення:  {s:.3f}')
print(f'Розмір вибірки:  {n}')
```

```python
# 3. Вибір типу тесту та проведення тесту
# так як розмір вибірки менше 30, а стандартне відхилення популяції не відоме, тому застосовуємо t-тест Стьюдента

# t-тест
res = stats.ttest_1samp(times, popmean=mu0, alternative="greater")

# Критичне значення для правостороннього тесту
df = n-1
t_critical = stats.t.ppf(1-alpha,df)

print(f"t-test: t_statistic = {res.statistic:.3f}, t_critical = {t_critical:.3f}, p-value = {res.pvalue:.5f}")
```

```python
# 4. Прийняття рішення про гіпотези використовуючи p-value
if res.pvalue < alpha:
    print(f"Відхиляємо H_0: середній час до покупки користувачів більший за {mu0} на рівні значущості {alpha}")
else:
    print(f"Не відхиляємо H_0: на рівні значущості {alpha} доказів недостатньо")
```

5. Висновок

 Середній час до покупки після редизайну статистично значуще зріс на рівні значущості 10% (x̄ = 132.3 хв проти 123.7 хв, p = 0.042). Проте вибірка дуже мала (n = 7), тому стверджувати, що саме редизайн спричинив зростання, можна лише з обережністю.

```python
# 6. Зміна рівня значущості до 0.05
alpha_2 = 0.05
if res.pvalue < alpha_2:
    print(f"Відхиляємо H_0: середній час до покупки користувачів більший за {mu0} на рівні значущості {alpha_2}")
else:
    print(f"Не відхиляємо H_0: на рівні значущості {alpha_2} доказів недостатньо")
```

Висновок

Ні, рішення не зміниться. p-value = 0.042 все ще менше за рівень значущості 0.05.

Тому ми відхиляємо H₀ і на рівні значущості 5%: середній час до покупки статистично значуще збільшився.

Водночас запас невеликий: p-value близьке до 0.05, і при суворішому рівні значущості (наприклад, α = 0.04 ми вже не змогли б відхилити H₀. Разом з малою вибіркою (n = 7) це означає, що результат варто перевірити на більшій кількості користувачів.


**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
```

В даному випадку маємо справу з пропорціями, тому застосовуємо z - тест для пропорцій (через біноміальний розподіл)

Формулювання гіпотез:

H_0: p = 23% (упізнаваність логотипу компанії не зросла)

H_1: p > 23% (упізнаваність логотипу компанії зросла)

Тип тесту: правосторонній

Альтернативна гіпотеза ">”, тому тест правосторонній. Ми перевіряємо чи % впізнаваності саме збільшився, а не просто змінився.


```python
# дані з умови

count = 311    # кількість "успіхів"
nobs = 1200    # обсяг вибірки
p0 = 0.23      # гіпотетична пропорція (H_0)
a = 0.03       # рівень значущості
```

```python
# z-тест для пропорції
stat, pval = proportions_ztest(count, nobs, p0, alternative='larger')

print("🔹 Тест для однієї пропорції")
print(f"Z-статистика = {stat:.3f}")
print(f"p-value = {pval:.4f}")
```

```python
# Прийняття рішення про гіпотези
if pval < a:
    print(f"Відхиляємо H_0: рівень упізнаваності логотипу компанії більший за {p0*100}% на рівні значущості {a}")
else:
    print(f"Не відхиляємо H_0: на рівні значущості {a} доказів того, що впізнаваність більша за {p0*100}%, недостатньо")
```

Висновок:

На рівні значущості 3% є достатні докази, що впізнаваність логотипу зросла: у вибірці її показали 25.9% опитаних, p-value = 0.0106 < 0.03.

```python
# довірчий інтервал

confint = (proportion_confint(count, nobs, alpha=a, method='normal', alternative='smaller'))
print(f"{(1-a)*100}% довірчий інтервал: {confint}")
```

Оскільки тест правосторонній, використовуємо односторонній 97% довірчий інтервал [0.2354, 1]

Значення 23% з нульової гіпотези не потрапляє в інтервал, тому ми відхиляємо H₀. Це узгоджується з p-value = 0.0106 < 0.03.

Інтервал додатково показує величину ефекту: з впевненістю 97% впізнаваність логотипу після маркетингової кампанії становить щонайменше 23.54%, при точковій оцінці - 25.9%. Тобто зростання статистично значуще.
